# cutLPK: Quick Start & Experiment Reproduction

This notebook serves as a comprehensive guide to building, testing, and using **cutLPK**. It also includes scripts to reproduce experiments from the paper.

### **Notebook Contents:**

1.  **Build Dependencies**: Instructions to clone and build `cuPDLPx`, a critical dependency for `cutLPK`.
2.  **Build cutLPK**: Step-by-step guide to compiling the `cutLPK` project with CMake, linking against Gurobi, Eigen, and cuPDLPx.
3.  **Basic Usage Verification**: Simple commands to verify the installation by running `cutLPK` on a test dataset.
4.  **Precompiled package**: Scripts to use (`cutLPK_package.tar.gz`) for easy deployment.
5.  **Reproduce Experiments**:
    *   **Fair Clustering**: Batch scripts to run experiments on multiple datasets using both Alpha-Fairness and Tau-Fairness constraints.
    *   **Spectral Clustering**: Commands to reproduce spectral clustering results on social network datasets.

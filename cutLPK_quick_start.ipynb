compute tau that achieves same clustering balance given alpha

```python
def calculate_tau(V_g, alpha, K):
    V_g = V_g[V_g > 0]
    r_g = V_g / np.sum(V_g)
    B1_den = np.max(np.minimum(r_g * alpha + (1 - alpha), r_g / alpha))
    if B1_den == 0: return np.nan
    B1 = (np.min(r_g) * alpha) / B1_den
    C = np.min(V_g) / np.max(V_g)
    tau_den = C - B1 * (1 - K)
    return B1 / tau_den if tau_den > 0 else np.nan
```

## Plotting

```python
import re
import pandas as pd
import matplotlib.pyplot as plt
import os
from pathlib import Path
from matplotlib.lines import Line2D
from matplotlib.ticker import StrMethodFormatter
import numpy as np

def read_log_file(filepath):
    if not os.path.exists(filepath):
        print(f"Error: Log file not found at '{filepath}'.")
        return None
    try:
        with open(filepath, 'r') as f:
            return f.read()
    except Exception as e:
        print(f"Error reading file '{filepath}': {e}")
        return None

def parse_log(log_content):
    data = {}
    current_n_clusters = None
    current_param = None
    in_table = False

    # Regex to capture header data (n_clusters and fairness_param)
    re_n_clusters = re.compile(r"n_clusters:\s*(\d+)")
    re_fairness_param = re.compile(r"fairness_param:\s*([\d\.]+)")

    re_iter_line = re.compile(
        r"^\s*#?\s*"
        # Group 1: Optional '+' and Iter number
        r"(\+?\s*\d+)"
        # r"^\s*(\+?\s*\d+)"
        # Skip LB & UB columns (2 columns of scientific/float notation)
        r"(?:\s+[\d.e\+\-]+){2}"
        # Group 2: Capture Gap (4th data column)
        r"\s+([\d.e\+\-]+)"
        # Group 3: Capture T (5th data column)
        r"\s+(\d+)"
        # Skip Active, Violated, Total, Solver(s) (4 columns of scientific/float/integer notation)
        r"(?:\s+[\d.e\+\-]+){4}"
        # Group 4: Capture Total(s) (10th data column - the cumulative time)
        r"\s+([\d.e\+\-]+)"
    )


    for line in log_content.splitlines():
        #print(line)
        match_n_clusters = re_n_clusters.search(line)
        if match_n_clusters:
            current_n_clusters = int(match_n_clusters.group(1))
            if current_n_clusters not in data:
                data[current_n_clusters] = {}

        match_param = re_fairness_param.search(line)
        if match_param:
            current_param = float(match_param.group(1))
            data[current_n_clusters][current_param] = []

        # Check for start/end of table
        if "Iter" in line and "Total(s)" in line:
            in_table = True
            continue
        if re.match(r"-{10,}", line) or re.match(r"={10,}", line):
            continue
        if line.strip().startswith("Fair Lloyd objective"):
            # End of the relevant table output
            in_table = False

        # Process data line
        if in_table and current_n_clusters is not None and current_param is not None:
            match_iter_data = re_iter_line.match(line)
            if match_iter_data:
                # Groups: [Iter(+), Gap, T, Total(s)]
                iter_str = match_iter_data.group(1).strip()
                # Check for '+' prefix (used for T increase in the original log)
                is_t_increase = iter_str.startswith('+')

                try:
                    # Convert values to correct types
                    gap = float(match_iter_data.group(2))
                    t_val = int(match_iter_data.group(3))
                    total_time = float(match_iter_data.group(4))

                    data[current_n_clusters][current_param].append({
                        'Gap': gap,
                        'T': t_val,
                        'Total(s)': total_time,
                        'T_increase': is_t_increase
                    })
                except ValueError:
                    continue

    # Convert lists of dicts to DataFrames for easier manipulation
    final_data = {}
    for nc, params in data.items():
        final_data[nc] = {}
        for param, records in params.items():
            if records:
                df = pd.DataFrame(records)
                df['T_diff'] = df['T'].diff().fillna(0)
                df['T_marker'] = df['T_increase'] | (df['T_diff'] > 0)
                df['T_marker_shifted'] = df['T_marker'].astype(bool).shift(1).fillna(False)

                final_data[nc][param] = df

    return final_data
```

```python
def _apply_padded_ylim(ax, ylim, pad_frac: float = 0.10, hide_y_ticks: bool = True):
    lo, hi = ylim
    span = max(hi - lo, 1e-12)
    pad = span * max(pad_frac, 0.0)
    lo_p, hi_p = lo - pad, hi + pad
    ax.set_ylim(lo_p, hi_p)
    if hide_y_ticks:
        ax.tick_params(axis='y', which='both', length=0)
        ax.set_yticklabels([])
    return lo_p, hi_p


def plot_gap_trajectories(
    data_to_plot,
    n_clusters_list,
    params_list,
    filename_prefix,
    ylim=None,
    xlim=None,
    pad_frac=0.10,        # 10% padding above/below ylim
    show_y_ticks=True
):
    plt.rcParams.update({
        "font.family": "serif",
        "mathtext.fontset": "stix",
        "figure.figsize": (7, 5.5)
    })

    colors  = ['b', 'g', 'r', 'c', 'm', 'y']
    markers = ['o', 's', 'D', '^', 'v', 'P']

    figs = []

    for n_clusters in n_clusters_list:
        if n_clusters not in data_to_plot:
            print(f"No data for n_clusters={n_clusters}. Skipping.")
            continue

        fig, ax = plt.subplots()

        # Compute padded ylim first
        if ylim is not None:
            y_low, y_high = _apply_padded_ylim(ax, ylim, pad_frac=pad_frac, hide_y_ticks=not show_y_ticks)
        else:
            ax.relim(); ax.autoscale(axis='y')
            lo, hi = ax.get_ylim()
            y_low, y_high = _apply_padded_ylim(ax, (lo, hi), pad_frac=pad_frac, hide_y_ticks=not show_y_ticks)

        if xlim is not None:
            ax.set_xlim(*xlim)

        handles, labels = [], []

        all_params = sorted(data_to_plot[n_clusters].keys(), reverse=True)
        is_tau = False
        if all(isinstance(p, int) for p in params_list):
            selected_params = [all_params[i] for i in params_list if 0 <= i < len(all_params)]
            is_tau = True
        else:
            # otherwise interpret params_list as actual numeric values to plot
            selected_params = [p for p in all_params if p in params_list]


        for idx, param in enumerate(selected_params):
            if param not in data_to_plot[n_clusters]:
                continue
            df = data_to_plot[n_clusters][param].copy()

            df = df[(df['Gap'] >= y_low) & (df['Gap'] <= ylim[1])]
            if df.empty:
                continue

            color  = colors[idx % len(colors)]
            marker = markers[idx % len(markers)]
            step   = max(1, len(df)//12)

            line, = ax.plot(
                df['Total(s)'],
                df['Gap'],
                label=fr'$\alpha = {param}$' if not is_tau else fr'$\tau = {param:.3f}$',
                color=color,
                linewidth=2,
                marker=marker,
                markersize=8,
                markerfacecolor='none',
                markeredgewidth=1.5,
                markevery=slice(0, None, step)
            )
            handles.append(line); labels.append(line.get_label())

            inc = df[df['T_marker_shifted']]
            if not inc.empty:
                ax.scatter(
                    inc['Total(s)'],
                    inc['Gap'],
                    color='gold',
                    marker='h',
                    edgecolors='k',
                    s=120,
                    zorder=5,
                    label='_nolegend_'
                )

        ax.set_xlabel("Elapsed Time (seconds)", fontsize=18)
        ax.set_ylabel("$\delta_g$", fontsize=18)
        ax.tick_params(axis='x', labelsize=18)
        if show_y_ticks:
            ax.tick_params(axis='y', labelsize=18)

        # Legend (add T increase)
        if handles:
            t_leg = Line2D([0], [0], marker='h', color='gold',
                           markeredgecolor='k', markersize=10, linestyle='',
                           label='$T$ increase')
            ax.legend(handles + [t_leg], labels + [t_leg.get_label()],
                      fontsize=18, loc='upper right', frameon=True, shadow=True)

        fig.tight_layout()
        pdf = f"{filename_prefix}_n_clusters_{n_clusters}.pdf"
        try:
            fig.savefig(pdf, dpi=600, bbox_inches='tight')
            print(f"Saved: {pdf}")
        except Exception as e:
            print(f"Error saving {pdf}: {e}")

        figs.append(fig)

    plt.show()
    return figs
```

```python
LOG_FILEPATH = '/content/fair_clustering_results/fair_kmeans_HC_alpha_output.txt'

filename_prefix = Path(LOG_FILEPATH).stem

N_CLUSTERS_TO_PLOT = [3, 4, 5]
PARAMS_TO_PLOT = [0.99, 0.9, 0.8, 0.7]

log_content = read_log_file(LOG_FILEPATH)

if log_content:
    parsed_data = parse_log(log_content)

    plot_gap_trajectories(parsed_data, N_CLUSTERS_TO_PLOT, PARAMS_TO_PLOT, filename_prefix, ylim=(0.00, 0.03))

    print(f"\nSuccessfully processed data for {len(N_CLUSTERS_TO_PLOT)} separate plots.")
    print(f"Read data from: '{LOG_FILEPATH}'")
else:
    print("\nPlotting aborted due to file reading error.")

LOG_FILEPATH = '/content/fair_clustering_results/fair_kmeans_HC_tau_output.txt'

filename_prefix = Path(LOG_FILEPATH).stem

N_CLUSTERS_TO_PLOT = [3, 4, 5]
PARAMS_TO_PLOT = [0, 1, 2, 3]

log_content = read_log_file(LOG_FILEPATH)

if log_content:
    parsed_data = parse_log(log_content)

    plot_gap_trajectories(parsed_data, N_CLUSTERS_TO_PLOT, PARAMS_TO_PLOT, filename_prefix, ylim=(0.00, 0.03))

    print(f"\nSuccessfully processed data for {len(N_CLUSTERS_TO_PLOT)} separate plots.")
    print(f"Read data from: '{LOG_FILEPATH}'")
else:
    print("\nPlotting aborted due to file reading error.")
```
